```python

sns.countplot(data=data, x="loan_status")
plt.title("Loan Status Distribution")
plt.xlabel("Loan Status")
plt.ylabel("Number of Customers")
plt.show()
```

```python

plt.figure(figsize=(8, 4))

sns.histplot(data=data, x="annual_income", kde=True)

plt.title("Annual Income Distribution")
plt.show()
```

```python

plt.figure(figsize=(8, 4))

sns.histplot(data=data, x="loan_amount", kde=True)

plt.title("Loan Amount Distribution")
plt.show()
```

```python

plt.figure(figsize=(8, 4))

sns.histplot(data=data, x="credit_score", kde=True)

plt.title("Credit Score Distribution")
plt.show()
```

```python

numeric_df = data.select_dtypes(include="number")

plt.figure(figsize=(14, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Loan Dataset Correlation Heatmap")
plt.show()
```

```python

plt.figure(figsize=(8, 4))

sns.countplot(data=data, x="occupation_status")

plt.title("Customers by Occupation Status")
plt.xticks(rotation=30)
plt.show()
```
